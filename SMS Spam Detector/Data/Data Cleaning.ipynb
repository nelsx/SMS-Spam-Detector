```python
import pandas as pd
```

```python
# Load the dataset
df = pd.read_csv('sms_spam_dataset.csv')
df.head()
```

```python
df.info()
```

```python
df['label'].value_counts()
```

```python
# Check for missing values
df.isnull().sum()
```

No missing values found in the dataset.

```python
# Check for duplicates
df.duplicated().sum()
```

```python
# Remove duplicates
df = df.drop_duplicates()
```

```python
# Check for duplicates again
df.duplicated().sum()
```

```python
df = df.drop_duplicates().reset_index(drop=True)
```

```python
# Check for special characters in messages
import re
special_characters = df['message'].apply(lambda x: bool(re.search(r'[^a-zA-Z0-9\s]', x)))
special_characters.sum()
```

```python
# Check for leading/trailing whitespaces in messages
df['message'] = df['message'].str.strip()
```

```python
#lowercase all messages
df['message'] = df['message'].str.lower()
```

```python
#Save the cleaned dataset
df.to_csv('cleaned_sms_spam_dataset.csv', index=False)
```
